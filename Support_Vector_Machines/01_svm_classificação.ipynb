## Codando Suport Vector Machive (SVM) ou Máquina de Vetores de Suporte para Classificação usando Python

```python
import numpy as np 
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./data/mouse_viral_study.csv')
```

```python
df.head()
```

```python
sns.scatterplot(
    data=df, 
    x='Med_1_mL',
    y='Med_2_mL', 
    hue='Virus Present'
)

# Hiperplano (Linha 2D)

x = np.linspace(0,10,100)
m = -1
b = 11
y = m * x + b

plt.plot(x, y, color='black', label='Hiperplano')
```

```python
from sklearn.svm import SVC
```

```python
help(SVC)
```

```python
y = df['Virus Present']
x = df.drop(columns=['Virus Present'])
```

```python
model = SVC(kernel='linear', C=1000)
```

```python
model.fit(x, y)
```

```python
from src.svm_margin_plot import plot_svm_boundary
```

```python
plot_svm_boundary(model, x, y)
```

```python
model = SVC(kernel='linear', C=0.05)
model.fit(x, y)
```

```python
plot_svm_boundary(model, x, y)
```

```python
model = SVC(kernel='rbf', C=1, gamma='scale')
model.fit(x, y)
plot_svm_boundary(model, x, y)
```

```python
model = SVC(kernel='rbf', C=1, gamma='auto')
model.fit(x, y)
plot_svm_boundary(model, x, y)
```

```python
model = SVC(kernel='rbf', C=1, gamma=2)
model.fit(x, y)
plot_svm_boundary(model, x, y)
```

```python
model = SVC(kernel='sigmoid', C=1)
model.fit(x, y)
plot_svm_boundary(model, x, y)
```

```python
model = SVC(kernel='poly', C=0.05, degree=4)
model.fit(x, y)
plot_svm_boundary(model, x, y)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
svm = SVC()
param_grid = {
    'C': [0.01, 0.1, 1],
    'kernel': ['linear', 'rbf', 'poly', 'sigmoid']
}
```

```python
grid = GridSearchCV(svm, param_grid)
```

```python
grid.fit(x, y)
```
