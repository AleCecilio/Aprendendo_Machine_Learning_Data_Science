Trabalharemos com o clássico Conjunto de Dados Iris. O conjunto de dados de flores Iris, ou conjunto de dados Iris de Fisher, é um conjunto de dados multivariados introduzido pelo estatístico, eugenista e biólogo britânico Ronald Fisher em seu artigo de 1936, "The use of multiple measurements in taxonomic problems as a example of linear discriminant analysis" (O uso de múltiplas medidas em problemas taxonômicos como um exemplo de análise discriminante linear).

Detalhes completos: https://en.wikipedia.org/wiki/Iris_flower_data_set

```python
X = df.drop('species', axis=1)
y = df['species']
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=101)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
log_model = LogisticRegression(solver='saga', max_iter=5000)
```

```python
penalty = ['l1', 'l2', 'elasticnet', 'none']
l1_ratio = np.linspace(0, 1, 20)
C = np.logspace(0, 10, 20) 

param_grid = {
    'penalty': penalty,
    'C': C,
    'l1_ratio': l1_ratio
}
```

```python
grid_model = GridSearchCV(log_model, param_grid=param_grid, n_jobs=-1)
```

```python
grid_model.fit(X_train_scaled, y_train)
```

```python
from sklearn.metrics import accuracy_score   
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import ConfusionMatrixDisplay
```

```python
grid_model.best_params_
```

```python
y_pred = grid_model.predict(X_test_scaled)
```

```python
y_pred
```

```python
accuracy_score(y_test, y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import roc_curve, auc
```

```python
def plot_multiclass_roc(clf, X_test, y_test, n_classes, figsize=(5,5)):
    y_score = clf.decision_function(X_test)

    # structures
    fpr = dict()
    tpr = dict()
    roc_auc = dict()

    # calculate dummies once
    y_test_dummies = pd.get_dummies(y_test, drop_first=False).values
    for i in range(n_classes):
        fpr[i], tpr[i], _ = roc_curve(y_test_dummies[:, i], y_score[:, i])
        roc_auc[i] = auc(fpr[i], tpr[i])

    # roc for each class
    fig, ax = plt.subplots(figsize=figsize)
    ax.plot([0, 1], [0, 1], 'k--')
    ax.set_xlim([0.0, 1.0])
    ax.set_ylim([0.0, 1.05])
    ax.set_xlabel('False Positive Rate')
    ax.set_ylabel('True Positive Rate')
    ax.set_title('Receiver operating characteristic example')
    for i in range(n_classes):
        ax.plot(fpr[i], tpr[i], label='ROC curve (area = %0.2f) for label %i' % (roc_auc[i], i))
    ax.legend(loc="best")
    ax.grid(alpha=.4)
    sns.despine()
    plt.show()
```

```python
plot_multiclass_roc(grid_model, X_test_scaled, y_test, n_classes=3)
```
