#### Informações sobre o Conjunto de Dados fornecidas pelos autores originais:

Criamos um banco de dados de dígitos coletando 250 amostras de 44 escritores. As amostras escritas por 30 escritores são utilizadas para treinamento, validação cruzada e testes dependentes do escritor, enquanto os dígitos escritos pelos outros 14 são utilizados para testes independentes do escritor. Este banco de dados também está disponível no formato UNIPEN.

Utilizamos uma mesa digitalizadora sensível à pressão WACOM PL-100V, com tela LCD integrada e caneta sem fio. As áreas de entrada e de visualização estão localizadas no mesmo lugar. Conectada à porta serial de um PC baseado em processador Intel 486, ela nos permite coletar amostras de escrita manual. A mesa envia as coordenadas $x$ e $y$ e os valores do nível de pressão da caneta em intervalos de tempo fixos (taxa de amostragem) de 100 milissegundos.

Solicita-se aos escritores que escrevam 250 dígitos em ordem aleatória dentro de caixas com resolução de 500 por 500 pixels da mesa. Os participantes são monitorados apenas durante as primeiras telas de entrada. Cada tela contém cinco caixas, com os dígitos a serem escritos exibidos acima delas. Os participantes são instruídos a escrever apenas dentro dessas caixas. Caso cometam um erro ou não estejam satisfeitos com sua escrita, são orientados a limpar o conteúdo de uma caixa utilizando um botão na tela. Os dez primeiros dígitos são descartados, pois a maioria dos escritores não está familiarizada com esse tipo de dispositivo de entrada, mas os participantes não sabem disso.

SOURCE: https://archive.ics.uci.edu/ml/datasets/Pen-Based+Recognition+of+Handwritten+Digits

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaled_pixels = scaler.fit_transform(pixels)
```

```python
scaled_pixels
```

```python
from sklearn.decomposition import PCA
```

```python
pca_model = PCA(n_components=2)
```

```python
pca_pixels = pca_model.fit_transform(scaled_pixels)
```

```python
np.sum(pca_model.explained_variance_ratio_)
```

```python
plt.figure(figsize=(10,6),dpi=150)
sns.scatterplot(
    x=pca_pixels[:, 0],
    y=pca_pixels[:, 1],
    hue=digits['number_label'].values,
    palette='Set1'
)
plt.legend(loc=(1.05,0))
```

```python
pca_3 = PCA(n_components=3)
```

```python
pca_pixels_3 = pca_3.fit_transform(scaled_pixels)
```

```python
from mpl_toolkits import mplot3d
```

```python
plt.figure(figsize=(8,8),dpi=150)
ax = plt.axes(projection='3d')
ax.scatter3D(
    pca_pixels_3[:,0],
    pca_pixels_3[:,1],
    pca_pixels_3[:,2],
    c=digits['number_label'].values
);
```
