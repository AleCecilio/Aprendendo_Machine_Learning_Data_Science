## Codando KNN (K Nearest Neighbors) ou K Vizinhos mais Próximos em Python 


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./data/gene_expression.csv')
```

```python
df.head()
```

```python
len(df)
```

```python
sns.scatterplot(
    x='Gene One', 
    y='Gene Two', 
    data=df, 
    hue='Cancer Present',
    alpha=0.5,
    style='Cancer Present'
)

plt.xlim(2,6)
plt.ylim(4,8)
```

```python
sns.pairplot(df, hue='Cancer Present')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
x = df.drop('Cancer Present', axis=1)
```

```python
y = df['Cancer Present']
```

```python
train_x, test_x, train_y, test_y = train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
scaler = StandardScaler()
```

```python
scaler_x_train = scaler.fit_transform(train_x)
```

```python
scaler_x_test = scaler.transform(test_x)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
help(KNeighborsClassifier)
```

```python
knn_model = KNeighborsClassifier(n_neighbors=1)
```

```python
knn_model.fit(scaler_x_train, train_y)
```

```python
y_pred = knn_model.predict(scaler_x_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
confusion_matrix(test_y, y_pred)
```

```python
len(test_y)
```

```python
print(classification_report(test_y, y_pred))
```

## Parte 2

- Como escolher o valor de k ideal?

```python
from sklearn.metrics import accuracy_score
```

```python
test_error_rates = []

for k in range(1,30):
    knn_model = KNeighborsClassifier(n_neighbors=k)
    knn_model.fit(scaler_x_train, train_y)

    y_pred_test = knn_model.predict(scaler_x_test)

    test_error = 1 - accuracy_score(test_y, y_pred_test)

    test_error_rates.append(test_error)
```

```python
test_error_rates
```

```python
plt.plot(range(1,30), test_error_rates)
plt.ylabel('Error Rate')
plt.xlabel('K Neighbors')
```

###  PIPELINE --> GRIDSEARCH CV

```python
scaler = StandardScaler()
```

```python
knn = KNeighborsClassifier()
```

```python
knn.get_params().keys()
```

```python
operations =[('scaler', scaler),('knn', knn)]
```

```python
from sklearn.pipeline import Pipeline
```

```python
pipe = Pipeline(operations) 
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
k_values = list(range(1,20))
```

```python
k_values
```

```python
param_grid = {'knn__n_neighbors': k_values}
```

```python
full_cv_classifier = GridSearchCV(pipe, param_grid, cv=5, scoring='accuracy')
```

```python
full_cv_classifier.fit(train_x, train_y)
```

```python
full_cv_classifier.best_estimator_.get_params()
```

```python
full_pred = full_cv_classifier.predict(test_x)
```

```python
print(classification_report(test_y, full_pred))
```

```python
new_patient = pd.DataFrame(
    [[3.8, 6.4]],
    columns=train_x.columns
)
```

```python
full_cv_classifier.predict(new_patient)
```
